# Phase 6f Symbolic: Curvature from SymPy Metric

SymPy Matrix로 metric을 정의하면 Christoffel, Riemann, Ricci, Einstein 텐서를 **수식**으로 얻을 수 있다.

$$g = \text{Metric}(\mathtt{sp.diag}(\ldots),\ \text{coords},\ \text{signature})$$
$$\texttt{sym} = g.\text{symbolic}([t, r, \theta, \varphi]) \quad\Rightarrow\quad \texttt{sym.R} = 12H^2$$

```python
import sys; sys.path.insert(0, "/home/minjae/Minjae/IndexCalc")
import sympy as sp
from indexcalc import Metric
```

## 1. FLRW 우주론: $R = 12H^2$

```python
t, x, y, z = sp.symbols('t x y z')
H = sp.Symbol('H', positive=True)

g_flrw = sp.diag(-1, sp.exp(2*H*t), sp.exp(2*H*t), sp.exp(2*H*t))
g = Metric(g_flrw, "cartesian", signature=(3,1))
sym = g.symbolic([t, x, y, z])

print(f"R = {sym.R}")
print(f"\nChristoffel:")
sym.show("christoffel")
print(f"\nRicci tensor:")
sym.show("ricci")

# 값 대입
print(f"\nR(H=0.5) = {sym.R.subs(H, sp.Rational(1,2))}")

assert sp.simplify(sym.R - 12*H**2) == 0
print("\n✓ 통과")
```

## 2. Schwarzschild: 진공해 $R = 0$, $K = 48M^2/r^6$

```python
r = sp.Symbol('r', positive=True)
theta, phi = sp.symbols('theta phi')
M = sp.Symbol('M', positive=True)

f = 1 - 2*M/r
g_sch = sp.diag(-f, 1/f, r**2, r**2 * sp.sin(theta)**2)
g = Metric(g_sch, "spherical", signature=(3,1))
sym = g.symbolic([t, r, theta, phi])

print(f"R = {sym.R}")
print(f"K = {sp.simplify(sym.K)}")
print(f"\nChristoffel (비영):")
sym.show("christoffel")

print(f"\nK(M=1, r=5) = {float(sym.K.subs({M:1, r:5})):.10f}")
print(f"  해석: {48/5**6:.10f}")
```

## 3. 2-Sphere: $R = 2/r_0^2$

```python
r0 = sp.Symbol('r_0', positive=True)
g_s2 = sp.diag(r0**2, r0**2 * sp.sin(theta)**2)
g = Metric(g_s2, "spherical", dim=2)
sym = g.symbolic([theta, phi])

print(f"R = {sym.R}")
print(f"Gaussian curvature K = R/2 = {sp.simplify(sym.R/2)}")
print(f"\nChristoffel:")
sym.show("christoffel")

assert sp.simplify(sym.R - 2/r0**2) == 0
print("\n✓ 통과")
```
