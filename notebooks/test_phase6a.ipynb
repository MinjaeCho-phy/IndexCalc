# IndexCalc — Phase 6a 테스트

편미분 ∂_μ: PartialDeriv, Leibniz rule 전개.

```python
import sys, os
sys.path.insert(0, os.path.abspath(".."))

from indexcalc import (
    IndexSpace, Tensor, IndexRegistry, parse, to_latex,
    partial, expand_partial,
)
from IPython.display import display, Math
```

```python
spacetime = IndexSpace("spacetime", dim=4, indices="μνλρσ", metric="g")
reg = IndexRegistry()
reg.register(spacetime)
```

## 1. 기본 편미분

```python
# ∂_μ V^ν
V = Tensor("V", [spacetime.upper("ν")])
mu = spacetime.lower("μ")
partial(V, mu)
```

```python
# ∂_μ g_{νλ}
g = Tensor("g", [spacetime.lower("ν"), spacetime.lower("λ")])
partial(g, mu)
```

```python
# 2차 미분: ∂_ν ∂_μ V
partial(partial(V, mu), spacetime.lower("ν"))
```

## 2. Leibniz rule 전개

```python
# ∂_μ (A^ν B_λ) → (∂_μ A^ν) B_λ + A^ν (∂_μ B_λ)
A = Tensor("A", [spacetime.upper("ν")])
B = Tensor("B", [spacetime.lower("λ")])

expr = partial(A * B, mu)
expanded = expand_partial(expr)

display(Math(to_latex(expr) + r" \;=\; " + to_latex(expanded)))
```

```python
# ∂_μ (A^ν + C^ν) → ∂_μ A^ν + ∂_μ C^ν
C = Tensor("C", [spacetime.upper("ν")])

expr = partial(A + C, mu)
expanded = expand_partial(expr)

display(Math(to_latex(expr) + r" \;=\; " + to_latex(expanded)))
```

```python
# ∂_μ (3 A^ν) → 3 ∂_μ A^ν
expr = partial(3 * A, mu)
expanded = expand_partial(expr)

display(Math(to_latex(expr) + r" \;=\; " + to_latex(expanded)))
```

## 3. 물리학 예시: ∂_μ (g_{νλ} V^λ)

```python
# ∂_μ (g_{νλ} V^λ) = (∂_μ g_{νλ}) V^λ + g_{νλ} (∂_μ V^λ)
g = Tensor("g", [spacetime.lower("ν"), spacetime.lower("λ")])
Vl = Tensor("V", [spacetime.upper("λ")])

expr = partial(g * Vl, mu)
expanded = expand_partial(expr)

display(Math(to_latex(expr)))
display(Math(r"= " + to_latex(expanded)))
```

## 4. Christoffel symbol 재료: ∂g 항들

```python
# Γ^μ_νλ = ½ g^{μρ} (∂_ν g_{ρλ} + ∂_λ g_{ρν} - ∂_ρ g_{νλ})
# 각 항을 symbolic하게 구성

nu = spacetime.lower("ν")
lam = spacetime.lower("λ")
rho = spacetime.lower("ρ")

g_rl = Tensor("g", [spacetime.lower("ρ"), spacetime.lower("λ")])
g_rn = Tensor("g", [spacetime.lower("ρ"), spacetime.lower("ν")])
g_nl = Tensor("g", [spacetime.lower("ν"), spacetime.lower("λ")])

term1 = partial(g_rl, nu)
term2 = partial(g_rn, lam)
term3 = partial(g_nl, rho)

display(Math(r"\partial_\nu g_{\rho\lambda} = " + to_latex(term1)))
display(Math(r"\partial_\lambda g_{\rho\nu} = " + to_latex(term2)))
display(Math(r"\partial_\rho g_{\nu\lambda} = " + to_latex(term3)))
```
